Load Data Hasil UMAP

```python
import numpy as np
import pandas as pd

# 1. Baca file CSV langsung dari direktori utama Colab
df_umap = pd.read_csv("hasil_feature_engineering_umap.csv")

# 2. Ambil kolom fitur koordinat UMAP
X_umap = df_umap.select_dtypes(include=[np.number]).values

print("Ukuran X_umap:", X_umap.shape)
```

Hierarchical Clustering (Agglomerative)

```python
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

# A. Visualisasi Dendrogram
plt.figure(figsize=(10, 5))
dendrogram = sch.dendrogram(sch.linkage(X_umap, method="ward"))
plt.title("Dendrogram Hierarchical Clustering")
plt.xlabel("Indeks Data")
plt.ylabel("Jarak Euclidean")
plt.show()

# B. Modeling (misal ditentukan n_clusters = 3)
hc = AgglomerativeClustering(n_clusters=3, metric="euclidean", linkage="ward")
df_umap["cluster_hc"] = hc.fit_predict(X_umap)
```

Non-Hierarchical Clustering (K-Means)

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Modeling K-Means (misal n_clusters = 3)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_umap["cluster_kmeans"] = kmeans.fit_predict(X_umap)

# Visualisasi Hasil K-Means
plt.figure(figsize=(8, 5))
plt.scatter(
    X_umap[:, 0],
    X_umap[:, 1],
    c=df_umap["cluster_kmeans"],
    cmap="viridis",
    s=50,
    alpha=0.7,
)
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    s=200,
    c="red",
    marker="X",
    label="Centroid",
)
plt.title("Non-Hierarchical Clustering (K-Means)")
plt.legend()
plt.show()
```

Density-Based Clustering (DBSCAN)

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

# Modeling DBSCAN (-1 mewakili noise/outlier)
dbscan = DBSCAN(eps=0.5, min_samples=3)
df_umap["cluster_dbscan"] = dbscan.fit_predict(X_umap)

# Visualisasi Hasil DBSCAN
plt.figure(figsize=(8, 5))
plt.scatter(
    X_umap[:, 0],
    X_umap[:, 1],
    c=df_umap["cluster_dbscan"],
    cmap="tab10",
    s=50,
    alpha=0.7,
)
plt.title("Density-Based Clustering (DBSCAN)")
plt.show()
```

```python
# 1. Gabungkan hasil klaster dengan dataset utama
df_klub = pd.read_csv("daftar_klub_indonesia_processed.csv")

# 2. Masukkan label klaster dari ketiga metode ke dataframe
df_klub["Cluster_Hierarchical"] = df_umap["cluster_hc"]
df_klub["Cluster_KMeans"] = df_umap["cluster_kmeans"]
df_klub["Cluster_DBSCAN"] = df_umap["cluster_dbscan"]

# 3. Tampilkan tabel perbandingan (10 baris pertama)
display(
    df_klub[
        [
            "Cluster_Hierarchical",
            "Cluster_KMeans",
            "Cluster_DBSCAN",
        ]  # Sesuaikan kolom teks utama jika ada, misal: 'nama_klub'
    ].head(10)
)
```
